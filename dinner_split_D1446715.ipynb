```python
# 第一組測試：原始資料與設定
event_name = " WEEKEND DINNER  "
meal_amount = "NT$1,200"
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300
```

```python
# 1. 整理聚餐名稱：去除頭尾空白並轉小寫
clean_event_name = event_name.strip().lower()

print("整理後的聚餐名稱：", clean_event_name)
```

```python
# 2. 整理餐費與人數，並進行型別轉換
# 移除 "NT$" 與 ","
temp_meal = meal_amount.replace("NT$", "").replace(",", "")
# 將清理後的字串轉成 float (浮點數)，保留原始資料不變
float_meal_amount = float(temp_meal)

# 將字串人數轉成 int (整數)
int_people = int(people)

print("轉換後的餐費數值：", float_meal_amount)
print("轉換後的人數數值：", int_people)
```

```python
# 3. 分步計算費用
service_fee = float_meal_amount * service_rate                   # 服務費
total_amount = float_meal_amount + service_fee - discount        # 應付總額 = 原始餐費 + 服務費 - 折價券
cost_per_person = total_amount / int_people                      # 每人費用 = 應付總額 ÷ 分攤人數

# 4. 判斷是否超過預算 (返回布林值 True / False)
is_over_budget = cost_per_person > budget_per_person

print("服務費：", service_fee)
print("應付總額：", total_amount)
print("每人費用：", cost_per_person)
print("是否超過預算：", is_over_budget)
```

```python
# 5. 輸出費用摘要
print("--- 聚餐費用摘要 ---")
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{float_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{int_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
